```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.metrics import silhouette_score
from scipy.cluster.hierarchy import dendrogram, linkage
```

```python
df = pd.read_csv("OnlineRetail.csv", encoding="ISO-8859-1")
df.head()
```

```python
df = df.dropna(subset=["CustomerID"])
df["Amount"] = df["Quantity"] * df["UnitPrice"]
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"], format="%d-%m-%Y %H:%M")
```

```python
reference_date = df["InvoiceDate"].max()

rfm = df.groupby("CustomerID").agg({
    "InvoiceDate": lambda x: (reference_date - x.max()).days,
    "InvoiceNo": "count",
    "Amount": "sum"
})

rfm.columns = ["Recency", "Frequency", "Monetary"]
rfm.head()
```

```python
scaler = StandardScaler()
rfm_scaled = scaler.fit_transform(rfm)
```

```python
wcss = []

for i in range(2,8):
    model = KMeans(n_clusters=i, random_state=42)
    model.fit(rfm_scaled)
    wcss.append(model.inertia_)

plt.plot(range(2,8), wcss, marker="o")
plt.xlabel("Clusters")
plt.ylabel("WCSS")
plt.title("Elbow Method")
plt.show()
```

```python
kmeans = KMeans(n_clusters=3, random_state=42)
rfm["KMeans_Cluster"] = kmeans.fit_predict(rfm_scaled)

rfm.head()
```

```python
score = silhouette_score(rfm_scaled, rfm["KMeans_Cluster"])
print("Silhouette Score:", score)
```

```python
plt.scatter(rfm["Frequency"], rfm["Monetary"], c=rfm["KMeans_Cluster"])
plt.xlabel("Frequency")
plt.ylabel("Monetary")
plt.title("K-Means Clusters")
plt.show()
```

```python
linked = linkage(rfm_scaled, method="ward")

plt.figure(figsize=(10,5))
dendrogram(linked)
plt.title("Hierarchical Clustering Dendrogram")
plt.show()
```

```python
hc = AgglomerativeClustering(n_clusters=3)
rfm["HC_Cluster"] = hc.fit_predict(rfm_scaled)

rfm.head()
```

```python
plt.scatter(rfm["Frequency"], rfm["Monetary"], c=rfm["HC_Cluster"])
plt.xlabel("Frequency")
plt.ylabel("Monetary")
plt.title("Hierarchical Clusters")
plt.show()
```
